```python
# 1. Input Collection: Prompt for player's name
name = input("Enter the player's name: ")

# 2. Numeric Input Processing: Accept and convert to integer
games_played = int(input("Enter the number of games played: "))

# 3. Score Data Entry: Collect total score and store as integer
total_score = int(input("Enter the total score achieved: "))

# 4. Computation: Calculate average score using division
# We use a float for average to keep precision
average_score = total_score / games_played

# 5. Output Display: Present results in the specified format
print("\nExpected Output Format:")
print(f"Player: {name}")
print(f"Games Played: {games_played}")
print(f"Total Score: {total_score}")
print(f"Average Score: {average_score}")
```
